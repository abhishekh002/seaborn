```python
import seaborn as sns
import matplotlib.pyplot
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
planets = sns.load_dataset('tips')
```

```python
print(planets)
```

```python
sns.set_theme(style="ticks")
g =   sns.relplot(
        data = planets,
        x = "total_bill",
        y = "tip",
        size = "size",
        hue = "smoker",
        style = "smoker"
    )
```

```python
g.savefig("scatter_plot.png")
```

```python
 k =    sns.relplot(
            data = planets,
            x = "total_bill",
            y = "tip",
            kind = "line"
        )
```

```python
k.savefig("line_plots.png")
```

```python
x_line = [1,2,3,4,5,6,7,8,9]
y_sq = [i**2 for i in x_line]
f =  sns.relplot(
        x = x_line,
        y = y_sq,
        kind = "line"
    )
```

```python
f.savefig("line_saves")
```
