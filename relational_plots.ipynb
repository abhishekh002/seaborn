```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
diamonds= sns.load_dataset('diamonds')
```

```python
print(diamonds.head())
```

```python
#scatter plots
p = sns.relplot(
        data = diamonds,
        x = "depth",
        y = "table",
        kind = "line"
        
    )
p.savefig("line_plot_2.png")
```

```python
#line plots
sns.relplot(
    data = diamonds,
    x = "depth",
    y = "table",
    hue = "color",
    style = "cut"
)
```
